A base tem **2.500 pedidos** e **19 colunas**, sem valores ausentes e sem linhas duplicadas. Cada pedido tem um identificador único e cada cliente aparece uma única vez, então não há risco de contar o mesmo cliente duas vezes.

A tabela de resumo traz, para cada variável: `count` (número de registros), `mean` (média), `std` (desvio padrão, que mede a dispersão em torno da média), `min` e `max` (menor e maior valor), e os quartis 25%, 50% (mediana) e 75%.

Alguns pontos que valem a atenção:

- **Nota de NPS:** a média é de 4,38 e a mediana 4,4. A maioria dos clientes dá notas baixas.

## 3. Variável-alvo e régua de classificação do NPS

### O que faremos e por quê
A variável que queremos explicar é `nps_score`. Como ela é uma nota de 0 a 10, o NPS clássico a classifica em três grupos: **detratores** (notas baixas), **neutros** (notas intermediárias) e **promotores** (notas altas). Com esses grupos calculamos o **índice NPS**, que é a porcentagem de promotores menos a porcentagem de detratores.

Nesta base a nota aceita decimais, então os cortes são feitos **sem arredondar**: detrator é nota menor que 7, neutro é de 7 até 8,9 e promotor é 9 ou mais. Usamos essa régua única em todo o notebook para que todas as tabelas e gráficos falem a mesma língua.

```python
nps = df["nps_score"]
assert nps.notna().all(), "nps_score tem valores ausentes"
assert nps.between(0, 10).all(), "nps_score está fora da escala de 0 a 10"

def categoria_nps(nota):
    return np.select([nota < 7, nota >= 9], ["Detrator", "Promotor"], default="Neutro")

ORDEM_CATEGORIAS = ["Detrator", "Neutro", "Promotor"]
df["nps_categoria"] = pd.Categorical(categoria_nps(nps), categories=ORDEM_CATEGORIAS, ordered=True)

contagem = df["nps_categoria"].value_counts().reindex(ORDEM_CATEGORIAS)
percentual = contagem / len(df) * 100
print(pd.DataFrame({"n": contagem, "%": percentual}).round(2))
print(f"\nÍndice NPS = %Promotores − %Detratores = {percentual['Promotor'] - percentual['Detrator']:.2f}")
print(f"Nota média: {nps.mean():.2f} | Mediana: {nps.median():.2f}")
```

### Resultados e interpretação
A tabela mostra quantos clientes caem em cada grupo:
- **Detratores: 2.109 clientes (84,4%).** A grande maioria da base está insatisfeita.
- **Neutros: 281 clientes (11,2%).**
- **Promotores: 110 clientes (4,4%).** Quase ninguém recomenda espontaneamente.

O **índice NPS** é de **−80**. Ele varia de −100 (todos detratores) a +100 (todos promotores), então um valor de −80 indica uma situação muito ruim. Os números de nota média (4,4) e mediana (4,4) são quase iguais, o que sugere que a distribuição das notas não é muito enviesada no centro. Ainda assim, a concentração em notas baixas é evidente, e o gráfico da seção 4 mostra isso com mais clareza.

## 4. Normalidade da nota

### O que faremos e por quê
Testamos se a nota de NPS segue uma distribuição **normal** (o formato de sino, simétrico em torno da média). Usamos dois instrumentos:
- **Teste de Shapiro-Wilk:** verifica formalmente a hipótese de normalidade. O p-valor abaixo de 0,05 indica que a normalidade é rejeitada.
- **Q-Q plot:** compara os quantis observados com os quantis de uma normal teórica. Se os pontos seguem a reta vermelha, a distribuição é próxima da normal.

Esse diagnóstico é importante porque muitos testes clássicos (como a correlação de Pearson e o teste t) assumem normalidade. Se ela não se sustenta, optamos por métodos não paramétricos, que não dependem desse formato.

```python
stat_w, p_w = stats.shapiro(nps)
print(f"Shapiro-Wilk: W = {stat_w:.4f} | p-valor = {p_w:.4f}")
print("Normalidade rejeitada (p < 0,05)." if p_w < 0.05 else "Não há evidência contra a normalidade.")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(nps, bins=np.arange(0, 10.5, 0.5), kde=True, color="teal", ax=axes[0])
for corte in (7, 9):
    axes[0].axvline(corte, ls="--", color="grey")
axes[0].set_title("Distribuição da nota (linhas em 7 e 9)", fontweight="bold")
axes[0].set_xlabel("Nota de NPS (0 a 10)")

stats.probplot(nps, dist="norm", plot=axes[1])
axes[1].set_title("Q-Q plot contra a normal teórica", fontweight="bold")
plt.tight_layout()
plt.show()
```

### Resultados e interpretação
O Shapiro-Wilk retorna W = 0,98 e um p-valor extremamente pequeno. Como o p-valor é menor que 0,05, **rejeitamos a normalidade**. Um W próximo de 1 pode dar a impressão de que a distribuição é quase normal, mas com 2.500 observações até desvios pequenos se tornam estatisticamente detectáveis. Por isso, olhamos também os gráficos.

No **histograma**, a maior parte das notas se concentra entre 3 e 6, com as barras mais altas perto de 4 e 5. As linhas tracejadas em 7 e 9 mostram os cortes da régua: há uma parcela grande abaixo de 7 e uma parcela pequena acima de 9.

No **Q-Q plot**, os pontos seguem a reta no centro, mas se afastam nas extremidades. Nas notas baixas e altas, há platôs horizontais em 0 e em 10. Isso acontece porque a escala é **limitada**: a nota não pode ser menor que 0 nem maior que 10, então muitos clientes ficam exatamente nos extremos. Essa limitação, e não uma assimetria simples, é a principal causa do afastamento da normal.

**Conclusão para o restante do notebook:** a nota não é normal. Por isso usamos Spearman (que trabalha com postos) e testes não paramétricos nas seções seguintes.

## 5. Associação entre cada variável e a nota (Spearman)

### O que faremos e por quê
Queremos saber quais variáveis caminham junto com a nota. Para isso calculamos o **coeficiente de Spearman** (ρ) entre cada variável numérica e `nps_score`. Ele varia de −1 a +1:
- **Valor positivo:** quando a variável sobe, a nota tende a subir.
- **Valor negativo:** quando a variável sobe, a nota tende a cair.
- **Valor perto de zero:** não há associação monotônica relevante.

Escolhemos Spearman porque ele compara a **ordem** dos valores (postos), e não os valores em si. Assim, não depende de normalidade e não é distorcido por valores extremos, como o pedido de R$ 2.000.

No gráfico, as barras vermelhas são associações negativas, as verdes são positivas, e as cinzas são variáveis medidas junto ou depois da pesquisa, que não devem entrar no modelo.

```python
POS_PESQUISA = ["csat_internal_score", "repeat_purchase_30d"]
IDS = ["customer_id", "order_id"]

cols_num = [c for c in df.select_dtypes(include=np.number).columns if c not in IDS]
matriz_corr = df[cols_num].corr(method="spearman")

corr_nps = matriz_corr["nps_score"].drop("nps_score").sort_values()
print(corr_nps.round(4).to_string())

def cor_barra(variavel, rho):
    if variavel in POS_PESQUISA:
        return "#c9cedb"
    return "#e74c3c" if rho < 0 else "#2ecc71"

cores = [cor_barra(v, r) for v, r in corr_nps.items()]

fig, ax = plt.subplots(figsize=(10, 7))
ax.barh(corr_nps.index, corr_nps.values, color=cores)
ax.axvline(0, color="black", lw=1, ls="--")
ax.set_xlim(-1, 1)
ax.set_xlabel("Coeficiente de Spearman (ρ) com nps_score")
ax.set_title("Associação de cada variável com a nota de NPS", fontweight="bold", loc="left")
for i, v in enumerate(corr_nps.values):
    ax.text(v + (0.02 if v >= 0 else -0.02), i, f"{v:+.2f}",
            va="center", ha="left" if v >= 0 else "right", fontsize=9)
plt.tight_layout()
plt.show()
```

### Resultados e interpretação
As associações mais fortes são as de atraso, reclamações e contatos com o SAC, todas negativas:
- **Atraso na entrega (ρ = −0,59):** é a associação mais forte. Clientes com mais dias de atraso tendem a dar notas menores.
- **Número de reclamações (ρ = −0,49)** e **contatos com o SAC (ρ = −0,34):** também associações fortes e negativas.
- **Tempo de resolução (ρ = −0,19):** associação negativa mais fraca.

Frete, idade, tempo de relacionamento, prazo prometido, itens, desconto, tentativas de entrega, parcelas e valor do pedido têm valores entre −0,04 e +0,04. Isso significa que **essas variáveis não acompanham a nota** de forma relevante.

As duas barras cinzas, com ρ de +0,56 para a satisfação interna e +0,49 para a recompra, são as mais fortes do gráfico. Elas não entram no modelo porque são medidas junto ou depois da pesquisa: usá-las para prever a nota seria trapacear, pois a informação já contém o resultado.

**Limite importante:** o coeficiente mede associação, não causa. Atraso e reclamações podem andar juntos sem que um cause o outro, e o gráfico não permite distinguir isso.

### Matriz de correlação entre as variáveis

#### O que faremos e por quê
O gráfico anterior mostra a relação de cada variável com a nota. A matriz completa mostra também como as **variáveis se relacionam entre si**. Isso importa porque, se dois preditores forem muito parecidos, o modelo terá dificuldade em separar o efeito de cada um. Mostramos apenas a metade inferior da matriz, pois ela é simétrica.

```python
mascara = np.triu(np.ones_like(matriz_corr, dtype=bool))

plt.figure(figsize=(12, 10))
sns.heatmap(
    matriz_corr,
    mask=mascara,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    cbar_kws={"label": "ρ de Spearman"},
)
plt.title("Matriz de Spearman entre variáveis numéricas", fontsize=13, fontweight="bold", pad=15)
plt.tight_layout()
plt.show()
```

### Resultados e interpretação
Cada célula mostra a associação entre duas variáveis. Cores vermelhas indicam associação positiva, azuis indicam negativa, e cores claras indicam quase nenhuma relação.

Os pares que merecem atenção:
- **Contatos com o SAC × reclamações (ρ = 0,74):** são fortemente parecidos. Quem reclama muito também costuma acionar o SAC muito, então as duas variáveis carregam informação redundante.
- **Atraso × reclamações (ρ = 0,19) e atraso × contatos com o SAC (ρ = −0,02):** o atraso se associa de forma leve às reclamações, mas praticamente não se associa aos contatos.
- **Tempo de resolução × satisfação interna (ρ = −0,74):** associação forte, mas a satisfação interna é medida após a pesquisa e não pode ser usada como preditor.

A redundância entre contatos e reclamações é o ponto mais relevante para a modelagem: ao incluir as duas variáveis no mesmo modelo, uma pode "roubar" o efeito da outra. Por isso, a seção de modelagem usa apenas uma delas (contatos com o SAC).

## 6. Diagnóstico operacional: detratores e não detratores

### O que faremos e por quê
Queremos entender o que diferencia clientes insatisfeitos dos demais, do ponto de vista da operação. Para isso, dividimos a base em três grupos exclusivos, na seguinte ordem de prioridade:
1. **Detratores:** nota menor que 7.
2. **Não detratores com recompra:** neutros ou promotores que compraram de novo em 30 dias.
3. **Não detratores sem recompra:** neutros ou promotores que não compraram de novo.

Comparamos a média de cada indicador operacional entre os grupos. Como a recompra é medida depois da pesquisa, essa comparação serve para **mapear diferenças**, não para prever a nota. Por isso a usamos apenas de forma descritiva.

```python
INDICADORES = {
    "delivery_delay_days": "Atraso na entrega (dias)",
    "customer_service_contacts": "Contatos no SAC (qtd)",
    "resolution_time_days": "Tempo de resolução (dias)",
    "complaints_count": "Reclamações (qtd)",
    "order_value": "Valor do pedido (R$)",
    "freight_value": "Valor do frete (R$)",
}
SEG_DETRATOR = "Detratores"
SEG_RECOMPRA = "Não detratores com recompra"
SEG_SEM_RECOMPRA = "Não detratores sem recompra"
ORDEM_SEGMENTOS = [SEG_DETRATOR, SEG_RECOMPRA, SEG_SEM_RECOMPRA]

df["segmento_diagnostico"] = np.select(
    [df["nps_score"] < 7, df["repeat_purchase_30d"] == 1],
    [SEG_DETRATOR, SEG_RECOMPRA],
    default=SEG_SEM_RECOMPRA,
)
df["segmento_diagnostico"] = pd.Categorical(df["segmento_diagnostico"], categories=ORDEM_SEGMENTOS, ordered=True)

tabela = df.groupby("segmento_diagnostico", observed=True)[list(INDICADORES)].mean().T
tabela.insert(0, "Base total", df[list(INDICADORES)].mean())
tabela = tabela[["Base total"] + ORDEM_SEGMENTOS].rename(index=INDICADORES)

n_por_grupo = df["segmento_diagnostico"].value_counts().reindex(ORDEM_SEGMENTOS)
tabela.loc["n (clientes)"] = [len(df)] + n_por_grupo.tolist()
tabela
```

### Resultados e interpretação
A tabela tem uma coluna para a base inteira, usada como referência, e uma coluna para cada grupo. Cada número é a **média** do indicador no grupo. A última linha mostra quantos clientes há em cada grupo.

Comparando detratores com os não detratores sem recompra, a diferença é clara:
- **Atraso na entrega:** cerca de 2,4 dias para detratores, contra 1,2 para os não detratores sem recompra e 0,8 para os que recompraram.
- **Contatos com o SAC:** cerca de 1,6 contato por pedido entre detratores, contra 1,1 e 0,8 nos outros grupos.
- **Reclamações:** cerca de 4,4 entre detratores, contra 2,8 e 2,4.
- **Tempo de resolução:** cerca de 5,7 dias entre detratores, contra 4,7 e 4,1.

Já **valor do pedido** e **valor do frete** são parecidos entre os grupos (frete perto de R$ 38 em todos). Isso indica que o preço não separa os clientes insatisfeitos dos satisfeitos.

**Cuidado com a interpretação:** são diferenças de médias, sem teste estatístico formal. Elas indicam onde olhar, mas não provam que o atraso *causa* a insatisfação. Além disso, o grupo de recompra é definido por um evento posterior à pesquisa.

## 7. Testes de comparação entre grupos

### 7.1 Recompra em 30 dias (Mann-Whitney U)

#### O que faremos e por quê
A tabela anterior mostrou diferenças de médias entre quem recompra e quem não recompra. Agora testamos se essa diferença é estatisticamente significativa. Usamos o **teste de Mann-Whitney U**, que compara as distribuições de duas amostras independentes com base nos postos. Como a nota não é normal, é o teste adequado.

Também verificamos o quanto os dois grupos **se separam**: a maior nota de quem não recomprou e a menor nota de quem recomprou. Se não há sobreposição, o teste retorna U = 0 e o p-valor fica muito pequeno. Isso pode ser real, mas também pode indicar um problema na construção da variável, então vale checar.

```python
g_sem = df.loc[df["repeat_purchase_30d"] == 0, "nps_score"]
g_com = df.loc[df["repeat_purchase_30d"] == 1, "nps_score"]

print(f"Nota média sem recompra: {g_sem.mean():.2f} | com recompra: {g_com.mean():.2f}")
print(f"Maior nota sem recompra: {g_sem.max():.1f} | menor nota com recompra: {g_com.min():.1f}")
print(f"Distância entre os grupos: {g_com.min() - g_sem.max():.1f} ponto(s)")

stat_mw, p_mw = stats.mannwhitneyu(g_sem, g_com, alternative="two-sided")
print(f"\nMann-Whitney U = {stat_mw:.1f} | p-valor = {p_mw:.4f}")
```

### Resultados e interpretação
Quem recomprou tem nota média de 9,0, enquanto quem não recomprou tem 3,9. A maior nota de quem não recomprou é 7,9 e a menor de quem recomprou é 8,0, então os dois grupos **não se sobrepõem**: a distância é de apenas 0,1 ponto.

O Mann-Whitney retorna U = 0, que é o valor mínimo possível e corresponde a nenhuma sobreposição entre os grupos. O p-valor é muito menor que 0,001, então a diferença é estatisticamente significativa.

**Atenção:** uma separação tão completa é incomum em dados reais. Ela pode significar que a recompra e a nota foram geradas a partir de uma mesma lógica, ou que há um problema na forma como a base foi montada. Antes de usar esse resultado como evidência de que a satisfação leva à recompra, precisamos entender a origem dessas duas variáveis.

### 7.2 Região do cliente (Kruskal-Wallis)

#### O que faremos e por quê
Verificamos se a nota muda de uma região para outra. Como são cinco regiões, usamos o **teste de Kruskal-Wallis**, a versão não paramétrica da ANOVA. Ele testa a hipótese de que todas as regiões têm a mesma distribuição de notas. Se não houver diferença, o problema de satisfação é provavelmente de toda a operação, e não de uma região específica.

```python
grupos_regiao = [g["nps_score"].to_numpy() for _, g in df.groupby("customer_region")]
stat_kw, p_kw = stats.kruskal(*grupos_regiao)
print(f"Kruskal-Wallis H = {stat_kw:.4f} | p-valor = {p_kw:.4f}")
print("Não há evidência de diferença de nota entre regiões (p > 0,05)." if p_kw > 0.05
      else "Há evidência de diferença de nota entre regiões (p < 0,05).")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(x="repeat_purchase_30d", y="nps_score", data=df, hue="repeat_purchase_30d",
            palette=["#e74c3c", "#2ecc71"], legend=False, ax=axes[0])
axes[0].set_xticks([0, 1], ["Não recomprou", "Recomprou"])
axes[0].set_title("Nota por recompra em 30 dias\n(medida após a pesquisa)", fontweight="bold")
axes[0].set_xlabel("Recompra em 30 dias")
axes[0].set_ylabel("Nota de NPS")

sns.boxplot(x="customer_region", y="nps_score", data=df, hue="customer_region",
            palette="Set2", legend=False, ax=axes[1])
axes[1].set_title("Nota por região", fontweight="bold")
axes[1].set_xlabel("Região do cliente")
axes[1].set_ylabel("Nota de NPS")
axes[1].tick_params(axis="x", rotation=45)
plt.tight_layout()
plt.show()
```

### Resultados e interpretação
O Kruskal-Wallis retorna H = 3,38 com p-valor de 0,50. Como o p-valor é bem maior que 0,05, **não há evidência de diferença de nota entre as regiões**.

Nos **boxplots**, cada caixa mostra a distribuição das notas de um grupo: a linha no meio é a mediana, a caixa cobre os 50% centrais, e os traços mostram os extremos. Na figura da esquerda, a caixa verde (recompra) fica inteiramente acima da vermelha (sem recompra), o que é consistente com a separação encontrada no teste anterior. Na figura da direita, as cinco regiões têm caixas com posição e tamanho parecidos, o que confirma o resultado do Kruskal-Wallis.

**Leitura prática:** a insatisfação é um problema que aparece em todo o país. Isso aponta para processos padronizados da operação, e não para uma regional específica.

## 8. Incerteza da média da nota

### O que faremos e por quê
A média da nota é um número único, mas ela tem uma margem de incerteza. O **intervalo de confiança de 95%** é a faixa de valores dentro da qual a média verdadeira deve estar, com 95% de confiança. Calculamos esse intervalo de duas formas:
- **Pela distribuição t de Student:** a fórmula clássica, baseada no erro padrão da média.
- **Por bootstrap:** reamostramos a base com reposição 2.000 vezes e observamos a variação das médias. É um método que não depende de suposições sobre a distribuição.

Como a base inteira é a população deste estudo, o intervalo mostra a precisão com que conhecemos a média desses 2.500 pedidos. Se os dois métodos derem resultados parecidos, a conclusão não depende da escolha do método.

```python
nota = df["nps_score"].to_numpy()
media = nota.mean()

ic_t = stats.t.interval(0.95, df=len(nota) - 1, loc=media, scale=stats.sem(nota))

N_BOOT = 2000
reamostras = rng.choice(nota, size=(N_BOOT, len(nota)), replace=True)
medias_boot = reamostras.mean(axis=1)
ic_boot = np.percentile(medias_boot, [2.5, 97.5])

print(f"Média da nota: {media:.4f}")
print(f"IC 95% (t de Student): ({ic_t[0]:.4f}, {ic_t[1]:.4f})")
print(f"IC 95% (bootstrap, {N_BOOT} reamostras): ({ic_boot[0]:.4f}, {ic_boot[1]:.4f})")
```

### Resultados e interpretação
A média da nota é de 4,38. O intervalo pela t de Student vai de 4,28 a 4,48, e o bootstrap dá praticamente o mesmo intervalo, de 4,28 a 4,48.

Isso significa que a média verdadeira da base está, com 95% de confiança, entre 4,28 e 4,48. A margem é de apenas **0,1 ponto para cada lado**, o que é uma precisão alta para uma escala de 0 a 10. Como os dois métodos concordam, podemos confiar que a estimativa não depende de suposições de normalidade.

Na prática, o intervalo serve para afirmar que a média está na faixa dos detratores (abaixo de 7) com muita segurança.

## 9. Modelo preditivo

### 9.1 Ajuste da regressão linear (OLS)

#### O que faremos e por quê
Queremos estimar a nota de um cliente a partir de sinais operacionais que existem **antes da pesquisa**. Usamos uma **regressão linear múltipla** (OLS), que estima quanto cada preditor contribui para a nota, mantendo os outros constantes. Escolhemos esse modelo porque é simples de interpretar, e a gestão consegue entender o efeito de cada variável.

**Preditores escolhidos:**
- **Atraso na entrega:** a falha logística mais direta.
- **Contatos com o SAC:** medida de atrito no pós-venda.
- **Tempo de resolução:** duração do problema, quando ele existe.
- **Valor do pedido:** como controle, para ver se o preço muda a nota.

**Variáveis excluídas:** `repeat_purchase_30d` e `csat_internal_score`, que são medidas após a pesquisa.

**Ressalva:** ainda não sabemos se contatos com o SAC e tempo de resolução são registrados antes ou depois da pesquisa. Se forem posteriores, o modelo deve ser refeito sem eles. Esse ponto precisa ser confirmado com quem gerou os dados.

```python
FEATURES = [
    "delivery_delay_days",
    "customer_service_contacts",
    "resolution_time_days",
    "order_value",
]

X = df[FEATURES]
y = df["nps_score"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE
)

modelo = sm.OLS(y_train, sm.add_constant(X_train)).fit()
print(modelo.summary())
```

### Resultados e interpretação
A base foi dividida em 80% para treino (2.000 pedidos) e 20% para teste (500 pedidos). O modelo foi ajustado apenas com os dados de treino, e o conjunto de teste fica reservado para a validação na próxima seção.

**Como ler o sumário:**
- **R-squared (R²) = 0,529:** o modelo explica cerca de 53% da variação das notas no treino. Quanto mais próximo de 1, melhor o ajuste.
- **coef:** o efeito estimado de cada preditor. Ele diz quanto a nota prevista muda quando o preditor aumenta uma unidade, mantendo os demais fixos.
- **P>|t|:** o p-valor de cada coeficiente. Valores abaixo de 0,05 indicam que o efeito é estatisticamente diferente de zero.
- **[0.025, 0.975]:** o intervalo de confiança de 95% do coeficiente.

**Os coeficientes, um por um:**
- **Intercepto = 8,42:** é a nota prevista para um pedido sem atraso, sem contatos e sem tempo de resolução. Esse valor cai na faixa **neutra** (7 a 8,9), e não na faixa de promotor.
- **Atraso = −1,05:** cada dia de atraso está associado a uma queda de cerca de 1 ponto na nota prevista. É o efeito mais forte do modelo.
- **Contatos com o SAC = −0,76:** cada contato está associado a uma queda de cerca de 0,8 ponto.
- **Tempo de resolução = −0,15:** cada dia de resolução está associado a uma queda de 0,15 ponto. O efeito é pequeno.
- **Valor do pedido = +0,0005:** por real. Na prática, R$ 1.000 a mais estão associados a cerca de 0,5 ponto a mais na nota. É um efeito pequeno, embora estatisticamente significativo.

Todos os coeficientes têm p-valor de 0,000, então são estatisticamente diferentes de zero. Lembre-se de que são **associações**: o modelo não prova que reduzir atrasos aumentaria a nota.

### 9.2 Validação no conjunto de teste

#### O que faremos e por quê
Um modelo pode se ajustar bem aos dados de treino e falhar com dados novos. Por isso, avaliamos o modelo no **conjunto de teste**, que ele não viu durante o ajuste.

Comparamos o modelo com uma **linha de base**, que sempre prevê a média da nota do treino. Essa referência é importante: se o modelo não erra menos que a média, ele não tem utilidade preditiva. Usamos três métricas:
- **MAE (erro absoluto médio):** em média, quantos pontos a previsão erra.
- **RMSE (raiz do erro quadrático médio):** semelhante ao MAE, mas penaliza mais os erros grandes.
- **R²:** quanto da variação da nota no teste o modelo explica. Zero significa que o modelo não é melhor que a média.

Também verificamos quantas previsões saem da escala de 0 a 10. Como a nota não pode sair desse intervalo, limitamos as previsões a ele.

```python
y_pred = modelo.predict(sm.add_constant(X_test))
print(f"Previsões fora de [0, 10] antes do ajuste: {((y_pred < 0) | (y_pred > 10)).mean():.1%}")

y_pred_ajustado = y_pred.clip(0, 10)
print(f"Previsões na faixa de promotor (≥ 9): {(y_pred_ajustado >= 9).mean():.1%}")

baseline = np.full(len(y_test), y_train.mean())

def metricas(real, previsto):
    return {
        "MAE": mean_absolute_error(real, previsto),
        "RMSE": np.sqrt(mean_squared_error(real, previsto)),
        "R²": r2_score(real, previsto),
    }

comparacao = pd.DataFrame({
    "Modelo OLS": metricas(y_test, y_pred_ajustado),
    "Linha de base (média)": metricas(y_test, baseline),
})
comparacao
```

### Resultados e interpretação
A tabela compara o modelo com a linha de base no conjunto de teste:

| Métrica | Modelo OLS | Linha de base |
|---|---|---|
| MAE | 1,36 | 2,07 |
| RMSE | 1,71 | 2,51 |
| R² | 0,54 | 0,00 |

**Como ler:**
- **MAE de 1,36:** em média, a nota prevista pelo modelo erra cerca de 1,4 ponto. Prever sempre a média erraria cerca de 2,1 pontos. O modelo reduz o erro em cerca de um terço.
- **RMSE:** segue a mesma direção, com redução de 2,51 para 1,71.
- **R² de 0,54:** o modelo explica cerca de 54% da variação das notas no teste. A linha de base tem R² zero, como esperado por construção.

Isso mostra que o modelo **generaliza**: o desempenho no teste é quase igual ao do treino (R² de 0,53 no treino), então ele não está apenas decorando os dados.

**Duas observações sobre as previsões:**
- Cerca de **2,4%** das previsões saem de [0, 10] antes do ajuste. Por isso as limitamos ao intervalo válido.
- **Nenhuma previsão chega à faixa de promotor (≥ 9).** O modelo tende a prever notas entre detratores e neutros. Isso é coerente com a base, em que quase todos os clientes são detratores, e significa que o modelo serve melhor para identificar quem está em risco do que para identificar quem vai ser promotor.

## 10. Aplicação prática e limites

### O que os resultados permitem concluir
1. **A operação se associa fortemente à nota.** Atraso, reclamações e contatos com o SAC são as variáveis com associação mais forte. Preço e frete praticamente não se associam.
2. **Dá para estimar a nota com razoável precisão** a partir de sinais operacionais, com erro médio de cerca de 1,4 ponto e R² de 0,54 no teste, superior à linha de base.
3. **A base é muito insatisfeita.** Com 84% de detratores, qualquer ação para aumentar o número de promotores precisa partir de uma mudança grande, não de ajustes pequenos.

### Como usar o modelo, com cautela
A nota prevista pode ajudar a **priorizar** pedidos que merecem atenção, por exemplo os que já acumulam atraso ou vários contatos. Ela não substitui a pesquisa de NPS, que continua sendo a medida oficial. Como o modelo não prevê promotores, ele não deve ser usado sozinho para decidir ações de recompensa ou de incentivo a avaliações.

### Limites que precisam ser declarados
- **Associação não é causa.** Reduzir atrasos ou contatos com o SAC precisa ser testado em parte dos clientes antes de ser tratado como efeito garantido.
- **Momento do registro das variáveis.** Ainda não confirmamos se contatos e tempo de resolução são registrados antes da pesquisa. Se não forem, o modelo só deve usar atraso e valor do pedido.
- **Separação entre recompra e nota.** Enquanto a origem dessa variável não for esclarecida, a comparação de recompra serve como diagnóstico, não como prova.
- **Base de um único período.** A base não tem datas, então os resultados descrevem estes 2.500 pedidos e não necessariamente o comportamento futuro.
- **Faixas de ação não definidas.** O notebook não estabelece os cortes de risco que a operação usaria. Esses cortes devem ser definidos com dados adicionais e validados antes da aplicação.

### Próximos passos sugeridos
Confirmar com a equipe de dados quando cada variável é registrada, testar o modelo com pedidos de um período posterior e planejar um experimento controlado para verificar se reduzir atrasos muda de fato a nota.
